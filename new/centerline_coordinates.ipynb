```python
import pandas as pd

path_to_geometry_data = "/Users/bencampbell/code/county_coverage/data/cleaner/centerlines_geo.json"

coordinates = pd.read_json(path_to_geometry_data)
coordinates
```

```python
# get enclosing box for each coodinate list
inf = float(inf)

def get_box(coordinates):
    longitude_max = -inf
    longitude_min = inf
    latitude_max = -inf
    latitude_min = inf
    
    for longitude, latitude in coordinates:
        if longitude < longitude_min:
            longitude_min = longitude
        if longitude > longitude_max:
            longitude_max = longitude
        if latitude < latitude_min:
            latitude_min = latitude
        if latitude > latitude_max:
            latitude_max = latitude
    return {'longitude_min': longitude_min,
            'longitude_max': longitude_max, 
            'latitude_min': latitude_min,
            'latitude_max': latitude_max}

boxes = pd.DataFrame.from_dict(coordinates.GEOMETRY.apply(get_box).to_dict(), orient='index')
boxes
```

```python

pd.concat((coordinates, boxes), axis=1)
```
