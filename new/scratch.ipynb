```python

def splits(values, iterations=4):
    memo = list()
    if iterations == 1:
        for a, b in pairwise(values):
            memo.append(a)
            memo.append((a+b)/2)
        memo.append(b)
    else:
        for a, b in pairwise(splits(values, iterations=iterations-1)):
            memo.append(a)
            memo.append((a+b)/2)
        memo.append(b)
    return memo

splits([1,2],iterations=4)

def even_splits(iterations=4):
    denominator = 2 ** iterations
    return [n/denominator for n in range(denominator+1)]

even_splits(4)
```

```python
class Indexer:
    def __init__(self, low, high, precision=10):
        assert precision >= 1
        self.precision = precision
        self.denominator = 2**precision
        self.span = high - low
        self.low, self.high = low, high

    def scale_value(self, value):
        return value - self.low
        
    def parameterize_value(self, value):
        return (value - self.low)/self.span
    
    def index(self, value):
        if (value < self.low) or (value > self.high):
            return float("NAN")
        value = (value - self.low)/self.span
        denominator = self.denominator
        for num in range(1, denominator + 1):
            if value < num/denominator:
                return num - 1
            
class Indexer_2d:
    def __init__(self, long_min, long_max, lat_min, lat_max, precision=10):
        self.latitude_indexer = Indexer(lat_min, lat_max, precision=precision)
        self.longitude_indexer = Indexer(long_min, long_max, precison=precision)

```

```python
from math import floor


def quantize(n:float, precision=8, format_result=False):
    if n < 0:
        raise ValueError(f"Out of range. Too low: {n}")
    if n > 1:
        raise ValueError(f"Out of range: Too high: {n}")
    out = floor(n * (2 ** precision))
    if format_result:
        return f"{out:0{precision}b}"
    else:
        return out

quantize(1/2, 4, format_result=True)


class bits(int):
    def __new__(cls, value, *, size=None):
        if size is None:
            size = value.bit_length()
        else:
            size = max(value.bit_length(), size)
        out = super().__new__(cls, value)
        out._size = size
        return out





```
