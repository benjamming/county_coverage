```python
coordinate_pattern = re_compile("(-?[0-9]+\.[0-9]+),(-?[0-9]+\.[0-9]+)")
    
def string_to_coordinates(string):
    out = list()
    for longitude, latitude in coordinate_pattern.findall(string):
        out.append((float(longitude), float(latitude)))
    return out


def get_box_from_points(coordinates):
    longitude_min = float('inf')
    longitude_max = -float('inf')
    latitude_min = float('inf')
    latitude_max = -float('inf')
    # The default is set to +/- infinity to ensure that the 
    for longitude, latitude in coordinates:
        if longitude < longitude_min:
            longitude_min = longitude
        if longitude > longitude_max:
            longitude_max = longitude
        if latitude < latitude_min:
            latitude_min = latitude
        if latitude > latitude_max:
            latitude_max = latitude 
    return {'longitude_min':longitude_min, 'longitude_max':longitude_max,
            'latitude_min':latitude_min, 'latitude_max':latitude_max}
```
