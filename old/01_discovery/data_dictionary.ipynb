```python
import pyparsing as pp

import os
from os import path

#Sources
"~/code/county_coverage/data/raw/centerlines" 

centerlines_info_path = "../../data/raw/centerlines/centerlines.info"
os.getcwd()
```

```python
from string import ascii_uppercase, digits
col_chars = ascii_uppercase + "_" + digits
col_chars

column_name = pp.Word(col_chars)

number = pp.Word(pp.nums)

# literals
right_arr, up_arr, down_arr = (pp.Literal(A) for A in "►▲▼")
star = pp.Literal("*")
field_expr = (pp.Literal("Field") + column_name +

star + "Alias" + pp.match_previous_expr(column_name) +
star + "Data type" + pp.Word(pp.alphanums) +
star + "Width" + number() +
star + "Precision" + number() +
star + "Scale" + number() +
pp.Literal("Field description"))
```

```python

EOL = pp.LineEnd().suppress()

line = pp.LineStart() + pp.SkipTo(pp.LineEnd(), 
                                  failOn=pp.LineStart()+pp.LineEnd()) + EOL


lines = pp.OneOrMore(line)
field_expr = (field_expr + lines +
pp.Literal("Hide") + "Field" + pp.match_previous_expr(column_name) + up_arr)
```

```python
test = """Field COUNDIST
  ►
* Alias COUNDIST
* Data type SmallInteger
* Width 2
* Precision 5
* Scale 0
Field description
Louisville Metro Council district number: 1-26
Hide Field COUNDIST ▲"""

bullet = star +  ... + pp.LineEnd()
fx_data = pp.OneOrMore(bullet) + ...

fx = pp.Group(pp.Literal("Field") + column_name + pp.Literal("►")+ fx_data +
      "Hide Field" + pp.match_previous_expr(column_name) + up_arr)

r=fx.search_string(test)
r
```
